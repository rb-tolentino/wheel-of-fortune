```python
import random
import pandas as pd

class Wheel:
    """
    This class displays the wheel and gets a random value from it.
    """
    def __init__(self, wedges, round):
        self.wedges = wedges # wedges
        self.round = round # round number

    def spin(self):
        input("Enter any key/s to spin the wheel: ") # to spin wheel

        landed_on = random.choice(self.wedges)
        return landed_on

    def show(self):
        total = len(self.wedges)
        counts = {}
        for w in self.wedges:
            counts[w] = counts.get(w, 0) + 1

        print()
        print(f"{self.round}")
        for wedge, count in counts.items():
            percentage = (count / total) * 100
            print(f"{wedge} - {percentage:.2f}%")
        print()

    def replace_wild_card(self):
        self.wedges.remove("Wild Card")
        self.wedges.append("$500")

    def add_mystery_wedges(self):
        for i in range(2):
            self.wedges.insert(0, "Mystery Wedge")
            self.wedges.remove("$1000")

    def replace_mystery_wedge(self):
        self.wedges.remove("Mystery Wedge")
        i = self.wedges.index("$2500")
        self.wedges.insert(i+1, "$1000")

    def add_express_wedge(self):
        self.wedges.insert(0, "Express Wedge")
        self.wedges.remove("$700")

    def replace_express_wedge(self):
        self.wedges.remove("Express Wedge")
        self.wedges.append("$700")
```

```python
class Board:
    """
    This class displays the Board with the unknown phrase and category.
    It also checks if the player's guess is in the board or not.
    """
    def __init__(self, phrase, category, value = 0):
        self.phrase = phrase # phrase generated
        self.category = category # category of the phrase generated
        self.value = value # value generated from the spin
        self.guessed_letters = set() # set of guessed letters
        self.spaced_unknown_list = []

    def setup(self):
        self.spaced_unknown_list = [] # initialize a list of characters
        for char in self.phrase: # for every character in the phrase generated
           if char == " ": # if it's a space
              self.spaced_unknown_list.append(" ")  # add space to unknown list
           else: # if it's a character
              self.spaced_unknown_list.append("_") # add blank to unknown list

        self.spaced_unknown_string = " ".join(self.spaced_unknown_list) # join the unknown list to make it a string
        # print(self.phrase) # should be deleted in actual game, only for testing

    def show(self, guess=None):
        if guess: # if player entered a valid guess
            for i, char in enumerate(self.phrase):
                if char.upper() == guess.upper():
                    self.spaced_unknown_list[i] = char

        updated_string = ' '.join(self.spaced_unknown_list) # convert list back to string for displayed
        print(f"\nCategory: {self.category}") # print category
        print(updated_string) # print updated string
        print(f"Guessed letters: {', '.join(self.guessed_letters)}") # print guessed letters
        # print(self.phrase) # should be deleted in actual game, only for testing

    def check_consonant(self):

        consonant_tuple = ('B', 'C', 'D', 'F', 'G', 'H', 'J', 'K', 'L', 'M', 'N', 'P', 'Q', 'R', 'S', 'T', 'V', 'W', 'X', 'Y', 'Z')
        guessed_consonant = input("What consonant would you like to guess? Input: ").upper() # asks for a consonant

        if guessed_consonant in self.guessed_letters:
            while guessed_consonant in self.guessed_letters: # if already guessed that consonant, while so it keeps asking
                print(f"You already guessed {guessed_consonant}!") # display error message
                guessed_consonant = input("What consonant would you like to guess? Input: ").upper() # ask

        if guessed_consonant not in consonant_tuple:
            while guessed_consonant not in consonant_tuple: # if it is not a consonant, while so it keeps asking
                print("Not a consonant!") # display error message
                guessed_consonant = input("What consonant would you like to guess? Input: ").upper() # ask

        self.guessed_letters.add(guessed_consonant) # add guessed_consonant to guessed_letters

        if guessed_consonant in self.phrase: # if guessed_consonant is in phrase
            count = self.phrase.count(guessed_consonant) # it counts how many guessed_consonants are there

        else: # if not in phrase
            count = 0 # zero guessed_consonants

        return guessed_consonant, count # return the consonant and its count

    def check_vowel(self):

        vowel_tuple = ('A', 'E', 'I', 'O', 'U')
        guessed_vowel = input("What vowel would you like to guess? Input: ").upper() # asks for a vowel

        if guessed_vowel in self.guessed_letters:
            while guessed_vowel in self.guessed_letters: # if already guessed that vowel, while so it keeps asking for valid
                print(f"You already guessed {guessed_vowel}!") # display error message
                guessed_vowel = input("What vowel would you like to guess? Input: ").upper() # ask

        if guessed_vowel not in vowel_tuple:
            while guessed_vowel not in vowel_tuple: # if it is not a vowel, while so it keeps asking
                print("Not a vowel!") # display error message
                guessed_vowel = input("What vowel would you like to guess? Input: ").upper() # ask

        self.guessed_letters.add(guessed_vowel) # add guessed_vowel to guessed_letters

        if guessed_vowel in self.phrase: # if guessed_vowel is in phrase
            count = self.phrase.count(guessed_vowel) # it counts how many guessed_vowels are there

        else: # if not in phrase
            count = 0 # zero guessed_vowels

        return guessed_vowel, count # return the vowel and its count

    def check_solve(self):

        guessed_phrase = input("What phrase do you guess? Input: ").upper() # ask for guessed_phrase
        if guessed_phrase == self.phrase: # if it's the same phrase
            print("Correct!") # print success message
            self.spaced_unknown_list = list(self.phrase) # unknown list becomes list of self.phrase
            self.show() # show solved board
            return True
        else:
            print("Incorrect!") # incorrect guessed_phrase
            self.show()
            return False

    def guess_vowel(self): # only for express wedge
        guessed_vowel, count = self.check_vowel() # check if a vowel is on the board
        self.show(guessed_vowel) # show board with that guess
        return guessed_vowel, count

    def guess_consonant(self): # only for special wedges
        print("\nSince you landed on a special wedge, you are required to guess a consonant.")
        guessed_consonant, count = self.check_consonant() # check if a consonant is on the board
        self.show(guessed_consonant) # show board with that guess
        return guessed_consonant, count

    def vowel_consonant_or_solve(self, value, round_money = 0):
        while True: # Loop until a valid and affordable choice is made
            print(f"Would you like to buy a vowel for $250, guess a consonant to earn {value} each, or solve the puzzle?")
            vcs_input = input("\nInput 'buy', 'guess', or 'solve': ").lower() #asks if buy, guess, or solve

            if vcs_input == 'buy':
                if round_money < 250:
                    print("You do not have enough money to buy a vowel! Please choose 'guess' or 'solve'.")
                    continue # Re-prompt for input
                else:
                    guess, count = self.check_vowel() # check if guess is in board
                    self.show(guess) # show board with that guess
                    return vcs_input, count, guess, False

            elif vcs_input == 'guess':
                guess, count = self.check_consonant() # check if guess is in board
                self.show(guess) # show board with that guess
                return vcs_input, count, guess, False

            elif vcs_input == 'solve':
                return vcs_input, 0, 0, self.check_solve() # check if solve is correct

            else:
                print("Invalid input. Please input 'buy', 'guess', or 'solve'.")
                continue # Re-prompt for input
```

```python
class Player:
    """
    This class manages player attributes and actions.
    """

    players = []        # shared among all players
    turn_index = 0      # index of whose turn it is

    def __init__(self, name, secured_money = 0, round_money = 0, has_wild_card = False):
        self.name = name # player's name
        self.secured_money = secured_money # player's secured money
        self.round_money = round_money # player's round money
        self.has_wild_card = has_wild_card # if player has wild card...

        Player.players.append(self)  # register player

    def current(self):
        """Returns the player whose turn it is."""
        return Player.players[Player.turn_index]

    def end_turn(self):
        """Ends turn and moves to the next player, returning the new current player."""
        Player.turn_index = (Player.turn_index + 1) % len(Player.players)
        new_current_player = Player.players[Player.turn_index]
        print(f"\n{new_current_player.name}, it is now your turn!") # Print the name of the new current player
        return new_current_player

    def add_round_money(self, amount):
        """Adds money to the player's current round money."""
        self.round_money += amount

    def deduct_round_money(self, amount):
        """Deducts money from the player's current round money."""
        self.round_money -= amount

    def add_secured_money(self, amount):
        """Adds money to the player's secured money."""
        self.secured_money += amount

    def bank_current_round_money(self):
        """Moves current round money to secured money and resets round money."""
        self.secured_money += self.round_money
        self.round_money = 0

    def bankrupt(self):
        """Resets current round money to zero, and removes wild card if has."""
        self.round_money = 0
        if self.has_wild_card:
            self.lost_wild_card()

    def buy_vowel(self):
        """Deducts cost of a vowel from round money."""
        self.deduct_round_money(250)

    def receive_wild_card(self):
        """Gives the player a Wild Card."""
        self.has_wild_card = True
        print(f"{self.name} has received the Wild Card!")

    def uses_wild_card(self):
        """Removes the Wild Card from the player."""
        self.has_wild_card = False
        print(f"{self.name} uses Wild Card!")

    def lost_wild_card(self):
        """Player loses Wild Card due to landing on bankrupt."""
        self.has_wild_card = False
        print(f"{self.name} lost Wild Card!")
```

```python
class Game:
    """
    This class manages the state of the game, with various attributes and methods.
    """

    def __init__ (self, round_number):
        self.round_number = round_number
        self.amount_of_players = None

    def next_round(self):
        self.round_number += 1

    def generate_phrase_and_category(self, numbers_taken, puzzle_dictionary):
        """
        This function generates a random phrase and category.
        """
        random_number = random.randint(1, 1000) # generates a random number from 1 to 1000
        if random_number in numbers_taken: # if this number is already taken
            while random_number in numbers_taken: # while it's still there...
                random_number = random.randint(1, 1000) # keep asking for random number
        else: # if not yet taken
            numbers_taken.append(random_number) # add this random number to list of numbers taken

        random_phrase = puzzle_dictionary[random_number][0]
        random_category = puzzle_dictionary[random_number][1]
        return random_phrase, random_category

    def count_players(self):
        amount_of_players_valid = False
        while not amount_of_players_valid:
            try:
                aop = int(input("\nEnter amount of players (2-4): "))
                if aop >= 2 and aop <= 4:
                    amount_of_players_valid = True
                    self.amount_of_players = aop
                elif aop > 4:
                    print("Only a maximum of 4 players are allowed.")
                elif aop < 2:
                    print("Please enter at least 2 players.")
                else:
                    print("Please enter a valid number.")
            except ValueError:
                print("Invalid input. Please enter a number.")
        return self.amount_of_players

    def list_players(self):
        player_names = [] # Initialize as empty list to append valid names

        print("\nPlayer names must not exceed 10 characters.")

        for i in range(self.amount_of_players):
            name_valid = False
            while not name_valid:
                name = input(f"Enter player {i+1} name: ")
                if len(name) > 10:
                    print("Exceeded 10 characters.")
                elif len(name) == 0:
                    print("Name cannot be empty.")
                elif name in player_names:
                    print("That name is already taken. Please choose another.")
                else:
                    player_names.append(name) # Add valid name to list
                    name_valid = True

        # Create Player objects from the collected names
        for name in player_names:
            Player(name) # This automatically adds the player to Player.players

    def input_to_proceed(self):
        input("Input any key/s to proceed: ")

    def welcome_screen(self):
        print("Welcome to Wheel of Fortune!\n")
        print("ENTER R TO READ RULES")
        print("ENTER ANY OTHER KEY/S TO PLAY\n")
        rule_or_not = input("Input: ")
        if rule_or_not.lower() == 'r':
            self.rule_book()
        else:
            pass

    def score_board(self, round_number):

        if round_number <= 3:
            print(f"\nCURRENT SCORE:")
            for player in Player.players:
                print(f"{player.name} -- ${player.secured_money} + (${player.round_money})")
            print(f"|Values in parenthesis are money obtained during this round and is subject to bankruptcy.|\n")
            self.input_to_proceed()

        elif round_number == 4:
            print(f"\nCURRENT SCORE:")
            for player in Player.players:
                print(f"{player.name} -- ${player.secured_money}")
            self.input_to_proceed()

        else:
            print(f"\nFINAL SCORE:")
            for player in Player.players:
                print(f"{player.name} -- ${player.secured_money}")

            highest_score = max(player.secured_money for player in Player.players)

            players_with_highest_score = [
                player for player in Player.players
                if player.secured_money == highest_score
            ]

            if len(players_with_highest_score) == 1:
                print(f"\nThe final winner is {players_with_highest_score[0].name}.")
            else:
                names = ", ".join(player.name for player in players_with_highest_score)
                print(f"\nThe final winners are: {names}.")

            print("\nThank you for playing Wheel of Fortune!")

    def rule_book(self):

        print("""

THE RULEBOOK

OBJECTIVE: Gain the most money by the end of the game.

RULES:
1. Given that you do not land on lose a turn or bankrupt, you may buy a vowel, guess a consonant, or solve the puzzle.
    - To buy a vowel, $250 will be deducted from your current round balance. You may not use secured money from previous rounds to buy a vowel.
    - Guessing a consonant correctly will give you (value * amount of consonants) to your current round balance.
    - Solving the puzzle will secure money for all players in that round.
2. Landing on bankrupt only clears your current round balance. Secured money from previous rounds will not be affected.

FLOW:
  - Round 1 (Wild Card)
  - Round 2 (Mystery Wedge)
  - Round 3 (Express Wedge)
  - Round 4 (Speed-Up Round)

SPECIAL WEDGES

There are 24 wedges in the wheel. Special wedges are placed each round to give chance to players who may be behind.

ROUND 1: Wild Card
    For round 1, the wild card is placed on a $500 wedge. Landing on the wild card alone is not enough to take possession of it,
    you need to guess a consonant correctly. In later parts of the game, you may use the wild card to guess an extra consonant with
    the same value after guessing a correct one. However, it is not a secured item; landing on bankrupt AT ANY POINT in the game
    loses you the wild card. This stays on the board unless taken.

ROUND 2: Mystery Wedge
    For round 2, two mystery wedges are placed on the $1000 wedges. Landing on the mystery wedge requires you to guess a consonant.
    Guessing a correct consonant gives you the choice to flip the wedge. One wedge has BANKRUPT (lose earnings for current round, and
    skipped turn) whereas the other wedge has $10000 (added to current round winnings immediately). The player also has the option to
    not flip the wedge. This stays on the board unless flipped.

ROUND 3: Express Wedge
    For round 3, an express wedge is placed on a $700 wedge. Guessing a correct consonant on the express wedge allows
    you to either skip the express (spin the wheel again), hop aboard the Express (risk current express earnings by guessing another
    consonant for $1000 each), buy a vowel for $250, or solve the puzzle. Guessing a wrong letter while on the Express causes the player
    to lose his turn, and lose all Express winnings. This stays on the board unless hopped on.
    * The only way a player can secure his express winnings is by solving the puzzle without incorrect guesses.

The wedges stay on the board unless landed on.

For round 4, there is no special wedge; however, it will be the speed-up round!

ROUND 4: Speed-Up Round
    For round 4, the speed-up round commences. In this round, all the special wedges would be removed; the wheel would be spun only once.
    The value landed upon would be added per correct consonant guessed. Additionally, players will take turns to guess letters. They may
    either guess a consonant to earn money, or guess a vowel for free. They are only allowed to guess once
    then they would be given a chance to solve the puzzle before the next player gets to guess. The round ends after a player solves the puzzle
    correctly.

The winner of the game is the one who obtained the most amount of money by the end.
In case of a tie, there would be multiple winners.

    """
)

        self.input_to_proceed()

    def intro_to_game(self):

        p1 = Player.players[0]
        p2 = Player.players[1]

        if self.amount_of_players == 2:
            print(f"\nWelcome to Wheel of Fortune, {p1.name} and {p2.name}!")
        elif self.amount_of_players == 3:
            p3 = Player.players[2]
            print(f"\nWelcome to Wheel of Fortune, {p1.name}, {p2.name}, and {p3.name}!")
        else:
            p3 = Player.players[2]
            p4 = Player.players[3]
            print(f"\nWelcome to Wheel of Fortune, {p1.name}, {p2.name}, and {p3.name}!")
            print(f"\nWelcome to Wheel of Fortune, {p1.name}, {p2.name}, {p3.name}, and {p4.name}!")

        print("There are four rounds!")
        print("For the first three rounds, you may either buy a vowel for $250, guess a consonant, or solve the puzzle.")
        print("Whereas, the fourth round is the speed-up round.")
        print("Your goal is to win as much money as possible. That being said, are you ready?\n")

        self.input_to_proceed()

    def start_round(self, round_number, category):

        if round_number == 1:
            print("""
For Round 1, a wild card is placed on the board! It is valued at $500.
If you land on the wild card and guess a consonant correctly,
you will be able to use it to guess an extra consonant with the same value.
The player may use it anytime, but it is not a secured item.
Landing on bankrupt AT ANY POINT in the game loses you the wild card.""")

        elif round_number == 2:
            print("""
For Round 2, two mystery wedges have been added! It is valued at $1000.
If you guess a consonant correctly, you may either take the given amount,
or choose to flip the board. The two wedges have the respective values:
1. Bankrupt (lose earnings for current round, and skipped turn)
2. $10000 ($10000 added to current round winnings immediately)""")

        elif round_number == 3:
            print("""
For Round 3, An express wedge is added! Express wedges are valued at $1000.
If you guess a consonant correctly, you may either:
a. skip the express: spin the wheel again
b. hop aboard the Express: risk current express earnings by guessing another consonant (get $1,000 per correct consonant)
c. solve the puzzle to secure winnings.""")

        else: #round number == 4:
            print("""
For Round 4, the wheel would be spun only once! Afterwards, the players
would take turns to guess consonants for that given value. You may also
guess a vowel for free, but you do not gain money from it.

Additionally, all the special wedges are gone.""")

        print(f"\nAlright, everyone! For Round {round_number}, here is the word with the category {category}.")
```

```python

def main():

    import pandas as pd
    import random

    numbers_taken = [] # numbers / ID of words already taken
    possible_mystery = ["Bankrupt", "$10000"] # possible values of Mystery Wedge

    round_1_wedges = [ # wedges for Round 1
    "Wild Card",
    "$2500",
    "$1000", "$1000",
    "$800", "$800", "$800", "$800",
    "$700", "$700",
    "$650",
    "$600",
    "$500", "$500", "$500", "$500", "$500", "$500", "$500", "$500", "$500",
    "Lose a Turn",
    "Bankrupt", "Bankrupt"
    ]

    round_4_wedges = [ # wedges for Round 4 (static, speed up)
    "$2500",
    "$1000", "$1000",
    "$800", "$800", "$800", "$800",
    "$700", "$700",
    "$650",
    "$600", "$600", "$600",
    "$500", "$500", "$500", "$500", "$500", "$500", "$500", "$500", "$500", "$500"
    ]

    sheet_url = 'https://docs.google.com/spreadsheets/d/1ZXjGa3OSlMtAw9qZ-1iANEdZodLNBsmmd5eTwPWo8pk/edit?pli=1&gid=855454350#gid=855454350'
    csv_url = sheet_url.replace('/edit?pli=1&gid=', '/export?format=csv&gid=')
    df = pd.read_csv(csv_url) # let dataframe be the sheets i made
    puzzle_dictionary = df.set_index('PhraseCounter')[['Phrase', 'Category']].T.to_dict('list') # i AIed this part... but i assume it makes a dictionary with key: phrases and value: category

    round_number = 1 # initializes round number
    game = Game(round_number)

    game.welcome_screen()
    game.count_players()
    game.list_players()
    game.intro_to_game()

    current_player = Player.players[Player.turn_index]

    solved = False

    phrase, category = game.generate_phrase_and_category(numbers_taken, puzzle_dictionary)
    board = Board(phrase, category)
    board.setup()

    while not solved:

        while game.round_number < 4:

            if game.round_number == 1:
                game.start_round(game.round_number, category)
                wheel = Wheel(round_1_wedges, "Round 1 Wheel: ")

            elif game.round_number == 2:
                game.start_round(game.round_number, category)
                wheel = Wheel(wheel.wedges, "Round 2 Wheel: ")

            elif game.round_number == 3:
                game.start_round(game.round_number, category)
                wheel = Wheel(wheel.wedges, "Round 3 Wheel: ")

            board.show()
            wheel.show()
            print(f"{current_player.name}, it's your turn, spin the wheel!")
            value = wheel.spin()

            print(f"\n{current_player.name}, you landed on {value}!")

            if value == "Lose a Turn" or value == "Bankrupt": # if the value is lose a turn or bankrupt
                print("You lose your turn!") # lose turn message

                if value == "Bankrupt":
                    print("You also lose all current round earnings!") # bankrupt message
                    current_player.bankrupt() # placeholder for set player to bankrupt
                current_player = current_player.end_turn() #placeholder for player loses turn

            elif value == "Wild Card" or value == "Mystery Wedge" or value == "Express Wedge":
                guessed_consonant, count = board.guess_consonant()

                if value == "Wild Card":

                    if count == 0: # no letters on board
                        print(f"There are no {guessed_consonant}s on the board. You fail to get the Wild Card.")
                        current_player = current_player.end_turn()
                    else: #letters on board
                        print("\nYou got the Wild Card!")
                        print(f"Also, you get $500 per correct consonant guessed. Hence, ${count * 500} has been added to your current round score!")
                        wheel.replace_wild_card()
                        current_player.add_round_money(count * 500)
                        current_player.receive_wild_card()

                elif value == "Mystery Wedge":

                    if count == 0: # no letters on board
                        print(f"There are no {guessed_consonant}s on the board. The Mystery Wedge stays on the board.")
                        current_player = current_player.end_turn()

                    else: #letters on board

                        print(f"\nThere are {count} {guessed_consonant}s on the board! You may take ${count * 1000} this round.")
                        print("However, as you landed on the Mystery Wedge, you may either:")
                        print(f"1. Take the ${count * 1000}")
                        print(f"2. Flip the Mystery Wedge (Bankrupt: {(possible_mystery.count("Bankrupt") / len(possible_mystery)) * 100}% - $10000: {(possible_mystery.count("$10000") / len(possible_mystery)) * 100}%)") # print(f" {wedge} - {percentage:.2f}%")

                        mystery_wedge_decision_valid = False

                        while not mystery_wedge_decision_valid:
                            mystery_wedge_decision = input("Input '1' or '2': ")

                            if mystery_wedge_decision == "1":
                                print(f"\n${count * 1000} has been added to your current round score!")
                                current_player.add_round_money(count * 1000)
                                mystery_wedge_decision_valid = True

                            elif mystery_wedge_decision == "2":
                                print("\nYou flipped the mystery wedge... you got...")
                                mystery_wedge_value = random.choice(possible_mystery)
                                print(f"{mystery_wedge_value.upper()}!")
                                possible_mystery.remove(mystery_wedge_value)
                                wheel.replace_mystery_wedge()

                                if mystery_wedge_value == "$10000":
                                    print("$10000 has been added to your current round score!")
                                    current_player.add_round_money(10000)

                                else: #mystery_wedge_value == "Bankrupt":
                                    print("You lose all your current round winnings!")
                                    current_player.bankrupt()
                                    current_player = current_player.end_turn()

                                mystery_wedge_decision_valid = True

                            else:
                                print("\nInvalid input.")

                else: #value == "Express Wedge":

                    express_wedge_invitation_message = f"""
Since you guessed a letter correctly on the Express Wedge, you may either:
a. skip the express: spin the wheel again, take ${count * 1000}
b. hop aboard the Express: risk current express earnings by guessing another letter (get $1,000 per correct consonant, buy vowels for $250)
c. solve the puzzle
* A wrong guess while on the express loses all express earnings; you need to solve the puzzle without incorrect guesses in order to secure it.
"""
                    express_wedge_aboarded_message = """
Since you guessed a letter correctly on the express wedge, you may either:
a. guess a consonant for $1000 per correct letter
b. buy a vowel for $250
c. solve the puzzle and secure express winnings.
"""
                    if count == 0:
                        print(f"There are no {guessed_consonant}s on the board. The Express Wedge stays on the board.")
                        current_player = current_player.end_turn()

                    else:

                        print(f"\nThere {'is' if count == 1 else 'are'} {count} {guessed_consonant}{'' if count == 1 else 's'} on the board! You may take ${count * 1000} this round.")
                        print(express_wedge_invitation_message)

                        express_wedge_decision_valid = False

                        while not express_wedge_decision_valid:

                            express_wedge_decision = input("Input 'a' or 'b', or 'c': ").lower()

                            if express_wedge_decision == "a":

                                express_wedge_decision_valid = True
                                print("\nYou skipped the express!")
                                current_player.add_round_money(count * 1000)
                                print(f"${count * 1000} has been added to your current round winnings!")

                            elif express_wedge_decision == "b":
                                express_wedge_decision_valid = True

                                wheel.replace_express_wedge()

                                aboard_express = True
                                express_winnings = count * 1000

                                print(f"\nHop aboard the express, {current_player.name}!")

                                while aboard_express:

                                    print(express_wedge_aboarded_message)

                                    express_wedge_input_valid = False

                                    while not express_wedge_input_valid:
                                        express_wedge_input = input("Input 'a', 'b', or 'c': ").lower()

                                        if express_wedge_input == 'a':
                                            express_wedge_input_valid = True
                                            guessed_consonant, count = board.check_consonant()
                                            board.show(guessed_consonant)
                                            if count == 0:
                                                express_winnings = 0
                                                aboard_express = False
                                                print(f"Unfortunately, there is no {guessed_consonant} in the board. You lose all express winnings.")
                                                current_player = current_player.end_turn()
                                            else:
                                                express_winnings += (count * 1000)
                                                print(f"There {'is' if count == 1 else 'are'} {count} {guessed_consonant}{'' if count == 1 else 's'} on the board! ${count * 1000} has been added to your express winnings.")
                                                print(f"CURRENT EXPRESS WINNINGS: ${express_winnings}")

                                        elif express_wedge_input == 'b':
                                            express_wedge_input_valid = True
                                            guessed_vowel, count = board.check_vowel() # check if a vowel is on the board
                                            board.show(guessed_vowel)
                                            if count == 0:
                                                express_winnings = 0
                                                aboard_express = False
                                                print(f"Unfortunately, there is no {guessed_vowel} in the board. You lose all express winnings.")
                                                current_player = current_player.end_turn()
                                            else:
                                                 express_winnings -= 250
                                                 print(f"There {'is' if count == 1 else 'are'} {count} {guessed_vowel}{'' if count == 1 else 's'} on the board! $250 has been deducted from your express winnings.")
                                                 print(f"CURRENT EXPRESS WINNINGS: ${express_winnings}")

                                        elif express_wedge_input == 'c':
                                            express_wedge_input_valid = True
                                            if board.check_solve():
                                                print("You solved the puzzle!")
                                                current_player.add_round_money(express_winnings)
                                                for player in Player.players:
                                                     player.bank_current_round_money()
                                                game.next_round()
                                                if game.round_number == 2:
                                                    wheel.add_mystery_wedges()
                                                elif game.round_number == 3:
                                                    wheel.add_express_wedge()
                                                aboard_express = False
                                                solved = True
                                                phrase, category = game.generate_phrase_and_category(numbers_taken, puzzle_dictionary)
                                                board = Board(phrase, category)
                                                board.setup()
                                            else:
                                                print("Incorrect! You lose all your express winnings!")
                                                current_player = current_player.end_turn()
                                                express_winnings = 0
                                                aboard_express = False

                                        else: #invalid input
                                             print("Invalid input.")

                            elif express_wedge_decision == "c":
                                express_wedge_decision_valid = True
                                if board.check_solve():
                                    print("You solved the puzzle!")
                                    for player in Player.players:
                                        player.bank_current_round_money()
                                    solved = True
                                    game.next_round()
                                    if game.round_number == 2:
                                        wheel.add_mystery_wedges()
                                    elif game.round_number == 3:
                                        wheel.add_express_wedge()
                                    phrase, category = game.generate_phrase_and_category(numbers_taken, puzzle_dictionary)
                                    board = Board(phrase, category)
                                    board.setup()
                                else:
                                    print("Incorrect. Express Wedge stays on the board.")
                                    current_player = current_player.end_turn()
                            else:
                                print("Invalid input.")

            else: # value is a normal dollar wedge

                value_dollar = int(value.replace('$', ''))

                vcs, count, guess, solved_checker = board.vowel_consonant_or_solve(value, current_player.round_money)

                if vcs == "buy":

                    current_player.buy_vowel()

                    if count == 0: # no correct vowels
                        print(f"There are no {guess}s on the board. $250 has been deducted from your score this round.")
                        current_player = current_player.end_turn()

                    else:
                        print(f"There {'is' if count == 1 else 'are'} {count} {guess}{'' if count == 1 else 's'} on the board! $250 has been deducted from your score this round.")

                elif vcs == "guess":
                    if count == 0: # no correct consonants
                        print(f"\nSorry, {current_player.name}, there are no {guess}s on the board! Your turn ends here.")
                        current_player = current_player.end_turn()

                    else: # correct consonants
                        print(f"There {'is' if count == 1 else 'are'} {count} {guess}{'' if count == 1 else 's'} on the board! ${count * value_dollar} has been added to your score this round!")
                        current_player.add_round_money(count * value_dollar)

                        if current_player.has_wild_card:
                            game.score_board(round_number)
                            wild_card_input = input(f"\n{current_player.name}, do you want to use the Wild Card to earn {value} per consonant? Input 'y' if yes, input any other key if no: ").lower()
                            if wild_card_input == 'y':
                                current_player.uses_wild_card()
                                guessed_consonant, count = board.check_consonant()
                                board.show(guessed_consonant)
                                if count == 0:
                                    print(f"\nSorry, {current_player.name}, there are no {guess}s on the board! Your turn ends here.")
                                    current_player = current_player.end_turn()
                                else:
                                    print(f"There {'is' if count == 1 else 'are'} {count} {guess}{'' if count == 1 else 's'} on the board! ${count * value_dollar} has been added to your score this round!")
                                    current_player.add_round_money(count * value_dollar)

                else: # vcs == "solve":
                    if solved_checker:
                        print(f"\nWell done, {current_player.name}! You solved the puzzle!")
                        print(f"All round winnings will be secured, except for Wild Card.")

                        for player in Player.players:
                            player.bank_current_round_money()

                        game.next_round()
                        if game.round_number == 2:
                            wheel.add_mystery_wedges()
                        elif game.round_number == 3:
                            wheel.add_express_wedge()
                        solved = True
                        phrase, category = game.generate_phrase_and_category(numbers_taken, puzzle_dictionary)
                        board = Board(phrase, category)
                        board.setup()
                    else:
                        print(f"\nIncorrect! {current_player.name}, since you guessed wrongly, you lose your turn.")
                        current_player = current_player.end_turn()

            game.score_board(round_number)

    if game.round_number == 4:

        solved = False
        phrase, category = game.generate_phrase_and_category(numbers_taken, puzzle_dictionary)
        board = Board(phrase, category)
        board.setup()

        game.start_round(4, category)

        board.show()

        wheel = Wheel(round_4_wedges, "Round 4 Wheel: ")    # static wheel
        wheel.show()

        print(f"{current_player.name}, it's your turn, spin the wheel!")
        value = wheel.spin()
        value_dollar = int(value.replace('$', ''))

        if value_dollar < 1000:
            print(f"As the wheel landed on {value} which is less than $1000, $1000 will be added to the value for the speed-up round.")
            value_dollar += 1000

        print(f"Therefore, every consonant guessed correctly this round is worth ${value_dollar}.")

        while not solved:

            print(f"\nNOTE: Vowels may be called for free, but does not earn you money.")
            print(f"You may guess a consonant for ${value_dollar} per correct letter, or call a vowel for free.")

            speed_up_round_decision_valid = False

            while not speed_up_round_decision_valid:

                speed_up_round_decision = input(f"{current_player.name}, Input 'consonant' or 'vowel': ").lower()
                permission_to_solve = False

                if speed_up_round_decision == 'consonant':
                    speed_up_round_decision_valid = True
                    guessed_consonant, count = board.check_consonant()
                    board.show(guessed_consonant)
                    if count == 0:
                        permission_to_solve = False
                        print(f"Unfortunately, there is no {guessed_consonant} in the board.")
                        current_player = current_player.end_turn()
                        print(f"It is now {current_player.name}'s turn!")
                    else:
                        print(f"There {'is' if count == 1 else 'are'} {count} {guessed_consonant}{'' if count == 1 else 's'} on the board! ${count * value_dollar} has been added to your score this round.")
                        current_player.add_secured_money(count * value_dollar)
                        permission_to_solve = True

                elif speed_up_round_decision == 'vowel':
                    speed_up_round_decision_valid = True
                    guessed_vowel, count = board.check_vowel()
                    board.show(guessed_vowel)
                    if count == 0:
                        permission_to_solve = False
                        print(f"Unfortunately, there is no {guessed_vowel} in the board.")
                        current_player = current_player.end_turn()
                        print(f"It is now {current_player.name}'s turn!")
                    else:
                        print(f"There {'is' if count == 1 else 'are'} {count} {guessed_vowel}{'' if count == 1 else 's'} on the board!")
                        permission_to_solve = True

                else:
                    print("Invalid input.")

            game.score_board(4)

            if permission_to_solve:
                print(f"\n{current_player.name}, what phrase do you guess? Input: ")

                if board.check_solve():
                    print("You solved the puzzle!")
                    for player in Player.players:
                        player.bank_current_round_money()
                    game.next_round()
                    solved = True

                else:
                    print("Incorrect. You lose your turn.")
                    current_player = current_player.end_turn()
                    game.score_board(4)

                permission_to_solve = False

    game.score_board(5)

if __name__ == '__main__':
    main()
```
